# Regression Model Using Keras

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from keras.models import Sequential
from keras.layers import Dense
```

```python
df = pd.read_csv("House_Price.csv")
```

```python
df.head()
```

```python
# Splitting predictors and target
predictors = df.iloc[:, 1:]  # All columns except the first (features)
target = df.iloc[:, 0]  # The first column (target variable)
```

```python
# Splitting the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(predictors, target, test_size=0.2, random_state=42)
```

```python
# Scaling the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

### Model Building

```python
# Building the model
model = Sequential()

# Input and first hidden layer
model.add(Dense(8, activation='relu', input_shape=(X_train.shape[1],)))

# Second hidden layer
model.add(Dense(5, activation='relu'))

# Third hidden layer
model.add(Dense(5, activation='relu'))

# Output layer
model.add(Dense(1))  # Single neuron for regression output
```

```python
# Compiling the model
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
```

```python
# Training the model
model.fit(X_train, y_train, epochs=50, batch_size=16, validation_split=0.2)
```

```python
# Evaluating the model
loss, mae = model.evaluate(X_test, y_test)
print(f"Test Loss: {loss}, Test MAE: {mae}")
```

```python
# Predicting house prices
predictions = model.predict(X_test)
```
