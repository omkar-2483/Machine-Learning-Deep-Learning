# Regression Model Using Keras: part-D

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

```python
df = pd.read_csv("concrete_data.csv")
```

```python
df.head()
```

```python
# Splitting predictors and target
X = df.iloc[:, -1:]  # All columns except the last (features)
y = df.iloc[:, -1]  # The last column (target variable)
```

```python
# Normalize the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_normalized = scaler.fit_transform(X)
```

```python
# Define the function to build the model
def build_model():
    model = Sequential([
        Dense(10, activation='relu', input_dim=X_normalized.shape[1]),
        Dense(10, activation='relu'),
        Dense(10, activation='relu'),
        Dense(1)  # Output layer for regression
    ])
    model.compile(optimizer=Adam(), loss='mean_squared_error')
    return model
```

```python
# Initialize variables to store results
mse_list = []

# Repeat the process 50 times
for i in range(50):
    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X_normalized, y, test_size=0.3, random_state=i)

    # Build the model
    model = build_model()

    # Train the model
    model.fit(X_train, y_train, epochs=50, verbose=0, batch_size=32)

    # Evaluate the model
    y_pred = model.predict(X_test).flatten()
    mse = mean_squared_error(y_test, y_pred)
    mse_list.append(mse)
```

```python
# Calculate mean and standard deviation of the mean squared errors
mean_mse = np.mean(mse_list)
std_mse = np.std(mse_list)

print(f"Mean MSE: {mean_mse}")
print(f"Standard Deviation of MSE: {std_mse}")
```
