What is TensorFlow?
* TensorFlow is a software framework developed by Google for building and deploying machine learning models. Unlike traditional programming, where you explicitly tell the computer what to do, TensorFlow allows you to train models using data so that the algorithm can come up with its own rules to complete tasks. 
* It's commonly used for deep learning applications such as image recognition, speech recognition, and language translation. TensorFlow is open-source and flexible, making it a popular choice for various machine learning projects.

Keras
* Keras is a high-level programming toolkit that simplifies the process of building neural networks. It acts as a wrapper for TensorFlow, allowing you to create complex models with just a few lines of code. 
* While TensorFlow provides low-level flexibility, Keras makes common tasks easier and faster by abstracting many of the complexities. It's recommended to learn TensorFlow first to understand the underlying processes, and then use Keras for more efficient development.

"Why is it called TensorFlow?":
* TensorFlow is named for its design, which involves working with large data sets stored in multi-dimensional arrays called tensors.
* Operations on these tensors are defined in a computational graph, similar to a flow chart, showing how data flows from one operation to the next
* The name "TensorFlow" reflects this flow of data (tensors) through the system.

The train/test/evaluation flow in TensorFlow
* Training Phase: The model is trained using training data, where it learns to transform input data into the expected output.
* Testing Phase: The trained model is tested with a new set of data (testing data) to ensure it can predict the correct results for data it hasn't seen before.
* Evaluation Phase: The model is used to make predictions on new data, and its performance is evaluated.
This process ensures that the model generalizes well and doesn't just memorize the training data.

# Creating TensorFlow Model

### Load Data

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# load training and test dataset from csv
train_data = pd.read_csv("sales_data_training.csv", dtype=float)
test_data = pd.read_csv("sales_data_test.csv", dtype=float)

#pull out columns with X and y
X_train = train_data.drop('total_earnings', axis=1).values
Y_train = train_data[['total_earnings']].values

X_test = test_data.drop('total_earnings', axis=1).values
Y_test = test_data[['total_earnings']].values

# all data need to be scaled to a small range line 0 to 1 for a neural network to work well
# Create scaler for input and output
X_scaler = MinMaxScaler(feature_range=(0,1))
Y_scaler = MinMaxScaler(feature_range=(0,1))

# Scale both input and output for training
X_scaled_train = X_scaler.fit_transform(X_train)
Y_scaled_train = Y_scaler.fit_transform(Y_train)

# Its very important that train and test data are scaled with same scaler
X_scaled_test = X_scaler.fit_transform(X_test)
Y_scaled_test = Y_scaler.fit_transform(Y_test)

print(X_scaled_test.shape)
print(Y_scaled_test.shape)
print("Note: Y values were scaled by multiplying by {:.10f} and adding {:.4f}".format(Y_scaler.scale_[0], Y_scaler.min_[0]))
```

Both X and Y are vector here

### Define Model Structure

```python
# Define model parameters
learning_rate = 0.001
training_epochs = 100

# Define the model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(50, activation='relu', input_shape=(X_train.shape[1],)),
    tf.keras.layers.Dense(100, activation='relu'),
    tf.keras.layers.Dense(50, activation='relu'),
    tf.keras.layers.Dense(1)  # Output layer
])

# Compile the model
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
              loss='mse')
```

### Train and Evaluate Model

```python
# Train the model
model.fit(X_scaled_train, Y_scaled_train, epochs=training_epochs, verbose=1, validation_data=(X_scaled_test, Y_scaled_test))
```

```python
# Evaluate the model
final_training_loss = model.evaluate(X_scaled_train, Y_scaled_train, verbose=0)
final_testing_loss = model.evaluate(X_scaled_test, Y_scaled_test, verbose=0)
print(f"Final Training Loss: {final_training_loss}")
print(f"Final Testing Loss: {final_testing_loss}")
```

```python
# Make predictions
Y_predicted_scaled = model.predict(X_scaled_test)
Y_predicted = Y_scaler.inverse_transform(Y_predicted_scaled)

real_earnings = test_data['total_earnings'].values[0]
predicted_earnings = Y_predicted[0][0]

print(f"The actual earnings of Game #1 were ${real_earnings}")
print(f"Our neural network predicted earnings of ${predicted_earnings}")
```
